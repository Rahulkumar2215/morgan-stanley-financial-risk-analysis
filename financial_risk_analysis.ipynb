```python
#Task 3
#Build the account summary
account_summary=data.groupby('AccountID').agg(
    txn_count=('TransactionID','size'),
    avg_balance=('AccountBalance','mean'),
    net_inflow=('SignedAmount','sum'),
    avg_risk=('RiskScore','mean')

).reset_index()
print(account_summary.head())
print("\nShape",account_summary.shape)
```

```python
#assume txn b/w (1-3) as Low,(4-6) as medium,7+ as High
account_summary['ActivityLevel']=pd.cut(
    account_summary['txn_count'],
    bins=[0,3,6,float('inf')],
    labels=['Low','Medium','High']
)
print(account_summary['ActivityLevel'].value_counts())
print(account_summary.head())
```

```python
account_summary['BalanceTier']=pd.qcut(
    account_summary['avg_balance'],q=3,labels=['Low','Medium','High']

)
print(account_summary['BalanceTier'].value_counts())
print(account_summary.groupby('BalanceTier')['avg_balance'].agg(['min','max','count']))
```

```python
#threshold=75th percentile of net inflow
# top 25 % of accounts by net inflow
inflow_threshold=account_summary['net_inflow'].quantile(0.75)
high_inflow=account_summary[account_summary['net_inflow']>=inflow_threshold]
print(f"Net inflow 75th percentile threshold:{inflow_threshold:}")
print(high_inflow.sort_values('net_inflow',ascending=False).head(10))
```

```python
#High frequency low balance accounts
high_fre_low_bal=account_summary[
    (account_summary['ActivityLevel']=='High')&(account_summary['BalanceTier']=='Low')
]
print(f"High-frequency,low-balance accounts,accounts:{len(high_fre_low_bal)}")
print(high_fre_low_bal)
```

```python
#Assume Large withdrawal=Txn whose amount is in top 10% (>=90th percentile) of all withdrawal amounts

withdrawals=data[data['TransactionType']=='Withdrawal']
large_threshold=withdrawals['TransactionAmount'].quantile(0.90)
large_withdrawals=withdrawals[withdrawals['TransactionAmount']>=large_threshold]
large_wd_counts=large_withdrawals.groupby('AccountID').size().sort_values(ascending=False)
print("large_threshold",large_threshold)
print("large withdrawals",large_withdrawals)
print("large_wd_count",large_wd_counts)
```

```python
# standard deviation and mean of balance per account
volatility=data.groupby('AccountID')['AccountBalance'].agg(['mean','std']).reset_index()

#coff of variation
volatility['cv']=volatility['std']/volatility['mean']
#most volatile accounts
volatility_sorted=volatility.sort_values('cv',ascending=False)
print(volatility_sorted.head(10))

```

```python
#Anomaly detection using IQR on txn amount
q1=data['TransactionAmount'].quantile(0.25)
q3=data['TransactionAmount'].quantile(0.75)
iqr=q3-q1

#define the outlier boundaries
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

#flag txn outside those bounds
amount_anomalies=data[(data['TransactionAmount']<lower_bound) | data['TransactionAmount']>upper_bound]

print("Number of anomalous txn:",len(amount_anomalies))
print(amount_anomalies[['AccountID','TransactionType','TransactionAmount']])
```

```python
q1=account_summary['net_inflow'].quantile(0.25)
q3=account_summary['net_inflow'].quantile(0.75)
iqr=q3-q1

#define the outlier boundaries
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

#flag txn outside those bounds
inflow_anomalies=account_summary[(account_summary['net_inflow']<lower_bound) | account_summary['net_inflow']>upper_bound]

print("Number of anomalous txn:",len(inflow_anomalies))
print(inflow_anomalies[['net_inflow']])
```

```python
#highlight irregular/suspicious accounts
#combining multiple parameters into account_summary table

#overdraft accounts flag
account_summary['flag_overdraft']=account_summary['AccountID'].isin(overdraft_ids)

#Large withdrawal flag

large_wd_ids=large_withdrawals['AccountID'].unique()
account_summary['flag_large_withdrawal']=account_summary['AccountID'].isin(large_wd_ids)

#high volatility flag

account_summary=account_summary.drop(columns=['cv'],errors='ignore')
account_summary=account_summary.merge(volatility[['AccountID','cv']],on='AccountID',how='left')
cv_threshold=account_summary['cv'].quantile(0.75)
account_summary['flag_high_volatility']=account_summary['cv']>=cv_threshold

#High Risk Score
risk_threshold=account_summary['avg_risk'].quantile(0.75)
account_summary['flag_risk']=account_summary['avg_risk']>=risk_threshold

#combine all flags
flags_col=['flag_overdraft','flag_large_withdrawal','flag_high_volatility','flag_risk']
account_summary['risk_flag_count']=account_summary[flags_col].sum(axis=1)

#watchlist
watchlist=account_summary.sort_values('risk_flag_count',ascending=False)
print(watchlist[['AccountID','risk_flag_count']+flags_col].head(10))

print("\n How many accounts trip each no of flags")
print(account_summary['risk_flag_count'].value_counts().sort_index())
```

```python
# Median Txn count
median_txn=account_summary['txn_count'].median()
print("Median Txn count",median_txn)

#Split accounts into high-volume and low-volume
high_volume=account_summary[account_summary['txn_count']>median_txn]['avg_balance']
low_volume=account_summary[account_summary['txn_count']<=median_txn]['avg_balance']

print(f"high volume group:{len(high_volume)}accounts,avg balance={high_volume.mean()}")
print(f"low volume group:{len(low_volume)}accounts,avg balance={low_volume.mean()}")
```

```python
# Test high volume Txn Accounts have statistically higher avg. balances than low-volume Accounts.
from scipy import stats

#H0:high-volume balance <=low-volume balance(NOT higher)
#H1:high-volume balance >low-volume(IS higher)

t_stat,p_value=stats.ttest_ind(high_volume,low_volume,equal_var=False,alternative='greater')

print(f"T statistic:{t_stat:4f}")
print(f"one-tailed p-value:{p_value:.4f}")

if p_value<0.05:
  print("Result:SIGNIFICANT(P<0.05)-reject H0. High-volume accounts Do have statistically higher average balances.")
else:
  print("Result:NOt SIGNIFICANT(P>=0.05)-Fail to reject H0. No statistical evidence that high volume accounts have higher balances.")
```
